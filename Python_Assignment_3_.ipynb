While loop & Control Statements :

Number Guessing Game

Explanation

* Here Import random and use randint function to generate random integer between 1 and 10. This number is stored in the variable secret_number.

* attempts=3, which means the user has three chances to guess the correct number

* The while loop runs continues as long as the value is greater than zero
* The input() function asks the user to enter a number.

* Inside the while loop, conditional statements are used to check the user's guess they act according to the user's input and display the appropriate result.

* And also used control statements such as continue and break  to control the flow of the program.

```python
# number guessing game

import random

secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
    guess_num = int(input("Guess a number between 1 and 10: "))

    if guess_num < 1 or guess_num > 10:
        print("Number is out of the valid range")
        continue

    elif guess_num > secret_number:
        print("Too high. Try again.")
        attempts = attempts - 1

    elif guess_num < secret_number:
        print("Too low. Try again.")
        attempts = attempts - 1

    else:
        print("Congratulations! You guessed the correct number.")
        break
else:
    print("Better luck next time!")
```

Multiplication Table Generator

Explanation

Here, we created a multiplication table using a for loop and the range function. The for loop is used when we want to repeat a block of code multiple times. Here, range(1, 11) generates numbers from 1 to 10 because the ending value, 11, is excluded. The variable i takes each value one by one, and the multiplication result is displayed for each value. Here we entered the input value as 4, so multiplication table of 4 is displayed.

```python
number=int(input("Enter the number for which you want the multiplication table: "))

for i in range(1,11):
  print(number,"x",i,"=",i*number)
```

Function

BMI Calculator

Here, we created a user-defined function named calculate_bmi to calculate the Body Mass Index using weight and height. Weight and height are the parameters.The return statement sends the calculated BMI value back.Here,we call the function by passing 58 as the weight and 1.62 as the height.

```python
def calculate_bmi(weight, height):
  BMI = weight/height**2
  return BMI

calculate_bmi(58,1.62)
```
